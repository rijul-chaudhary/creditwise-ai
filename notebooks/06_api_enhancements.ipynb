## Feature Generation Strategy

The final machine learning models were trained on 135 engineered features. However, requiring users to manually provide all of these values would result in a poor user experience and an impractical application workflow.

To address this challenge, CreditWise AI adopts a multi-layer feature generation architecture.

### Layer 1: User-Provided Features

These are the inputs directly entered by the user through the application form.

Examples include:

- Age
- Gender
- Annual Income
- Employment Years
- Loan Amount
- Credit Amount
- Education Level
- Family Status
- Credit Score

These fields are intuitive, easy to obtain, and closely aligned with information commonly requested during real loan applications.

### Layer 2: Backend-Derived Features

Several model features can be generated automatically from user-provided information.

Examples:

- AGE_YEARS derived from Age
- EMPLOYMENT_YEARS derived from Employment Years
- OWN_CAR_AGE_MISSING derived from vehicle ownership information
- EXT_SOURCE_MEAN derived from generated external source indicators

These features require no additional user effort and are calculated internally by the backend.

### Layer 3: Simulated External Credit Bureau Features

The original Home Credit dataset contains highly influential external bureau variables:

- EXT_SOURCE_1
- EXT_SOURCE_2
- EXT_SOURCE_3

In real financial institutions, these values are typically obtained through credit bureau integrations and are not directly provided by customers.

Since such integrations are unavailable in this project, a numerical credit score will be used as a proxy.

The backend will transform the supplied credit score into simulated EXT_SOURCE values that approximate external creditworthiness indicators while preserving the predictive behavior of the original model.

### Benefits of this Architecture

This approach provides several advantages:

- Reduces user input complexity
- Preserves important model signals
- Maintains compatibility with the trained models
- Supports future recommendation systems
- Enables what-if simulation capabilities
- Allows future integration with real credit bureau APIs

By separating user inputs from engineered features, the system achieves a balance between usability and predictive performance.

## Credit Score to EXT_SOURCE Transformation

The original model relies heavily on external bureau variables such as EXT_SOURCE_1, EXT_SOURCE_2, and EXT_SOURCE_3.

Since these variables are unavailable in a public deployment setting, CreditWise AI generates synthetic EXT_SOURCE values using the customer's credit score.

The transformation follows a simple principle:

- Higher credit scores produce higher EXT_SOURCE values.
- Lower credit scores produce lower EXT_SOURCE values.
- Generated values remain within the same numerical range used during model training.

This approach preserves the predictive behavior of the original model while allowing end users to provide only a single, intuitive creditworthiness indicator.

```python
def credit_score_to_ext_sources(credit_score):
    
    credit_score = max(
    300,
    min(850, credit_score)
)

    score_normalized = (
    credit_score - 300
) / 550
    
    ext_source_1 = score_normalized * 0.95
    
    ext_source_2 = score_normalized * 0.98
    
    ext_source_3 = score_normalized
    
    ext_source_mean = (
        ext_source_1 +
        ext_source_2 +
        ext_source_3
    ) / 3
    
    return {
        "EXT_SOURCE_1": round(ext_source_1, 4),
        "EXT_SOURCE_2": round(ext_source_2, 4),
        "EXT_SOURCE_3": round(ext_source_3, 4),
        "EXT_SOURCE_MEAN": round(ext_source_mean, 4)
    }
```

### Limitation

In a production environment, a more sophisticated mapping strategy could be employed, such as percentile-based calibration, historical risk curves, or direct credit bureau integrations.

The current design preserves the key relationship learned by the model:

Higher credit score → Higher EXT_SOURCE values → Lower default risk.

## Enhanced Credit Score Calibration

The original machine learning models rely heavily on external bureau variables such as EXT_SOURCE_1, EXT_SOURCE_2, and EXT_SOURCE_3. These variables are among the most influential predictors of default risk but are not directly available to end users.

To support deployment without access to real credit bureau integrations, CreditWise AI generates synthetic EXT_SOURCE values using the customer's credit score.

### Credit Score Categories

The calibration follows widely used credit score classifications:

* Poor: 300–579
* Fair: 580–669
* Good: 670–739
* Very Good: 740–799
* Excellent: 800–850

Each category is assigned a realistic EXT_SOURCE range representing increasing levels of creditworthiness.

### Piecewise Calibration Strategy

Instead of applying a single linear transformation across the entire credit score range, the system uses a piecewise mapping approach.

Within each credit category:

* Higher scores produce higher EXT_SOURCE values.
* Lower scores produce lower EXT_SOURCE values.
* Linear interpolation preserves score granularity.

This allows two customers within the same category to receive different creditworthiness estimates while maintaining realistic risk bands.

### Simulating Multiple Bureau Signals

The original dataset contains three independent bureau-derived variables:

* EXT_SOURCE_1
* EXT_SOURCE_2
* EXT_SOURCE_3

To better approximate real-world behavior, all three values are generated from the same underlying credit score but include small deterministic offsets.

This preserves the strong correlation expected between bureau signals while avoiding unrealistic identical values.

### Benefits

This approach provides:

* Realistic credit score interpretation
* Consistent and reproducible predictions
* Improved approximation of external bureau information
* Better alignment with real-world lending practices
* Compatibility with the original machine learning models

The resulting EXT_SOURCE variables serve as a practical substitute for unavailable bureau data while preserving the predictive behavior learned during training.


## Credit Score to EXT_SOURCE Generation

The following function converts a user's Credit Score (CIBIL Score) into synthetic bureau features.

The implementation combines:

- Industry-standard credit score categories
- Piecewise calibration
- Linear interpolation within each category
- Small deterministic offsets between bureau variables

This produces realistic EXT_SOURCE values while maintaining deterministic and reproducible behavior.

```python
def credit_score_to_ext_sources(credit_score):

    credit_score = max(
        300,
        min(850, credit_score)
    )

    bands = [
    (300, 579, 0.05, 0.30, "Poor"),
    (580, 669, 0.30, 0.55, "Fair"),
    (670, 739, 0.55, 0.75, "Good"),
    (740, 799, 0.75, 0.90, "Very Good"),
    (800, 850, 0.90, 1.00, "Excellent")
]

    for score_min, score_max, ext_min, ext_max, category in bands:

        if score_min <= credit_score <= score_max:

            position = (
                credit_score - score_min
            ) / (
                score_max - score_min
            )

            base_score = (
                ext_min +
                position * (ext_max - ext_min)
            )

            ext_source_1 = max(
                0,
                min(1, base_score * 0.98)
            )

            ext_source_2 = max(
                0,
                min(1, base_score)
            )

            ext_source_3 = max(
                0,
                min(1, base_score * 1.02)
            )

            ext_source_mean = (
                ext_source_1 +
                ext_source_2 +
                ext_source_3
            ) / 3

            return {
                "EXT_SOURCE_1": round(ext_source_1, 4),
                "EXT_SOURCE_2": round(ext_source_2, 4),
                "EXT_SOURCE_3": round(ext_source_3, 4),
                "EXT_SOURCE_MEAN": round(ext_source_mean, 4),
                "CREDIT_PROFILE_CATEGORY": category
            }
```

```python
for score in [
    300,
    575,
    580,
    650,
    700,
    750,
    850
]:
    
    print(
        f"\nCredit Score: {score}"
    )
    
    print(
        credit_score_to_ext_sources(score)
    )
```

Additionally, each generated bureau profile is assigned a Credit Profile Category (Poor, Fair, Good, Very Good, or Excellent). This category is derived directly from the user's credit score and serves as an interpretable representation of overall creditworthiness for frontend visualizations, recommendation systems, and what-if simulations.
