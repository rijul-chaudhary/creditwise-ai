## Credit Score to EXT_SOURCE Transformation

The original model relies heavily on external bureau variables such as EXT_SOURCE_1, EXT_SOURCE_2, and EXT_SOURCE_3.

Since these variables are unavailable in a public deployment setting, CreditWise AI generates synthetic EXT_SOURCE values using the customer's credit score.

The transformation follows a simple principle:

- Higher credit scores produce higher EXT_SOURCE values.
- Lower credit scores produce lower EXT_SOURCE values.
- Generated values remain within the same numerical range used during model training.

This approach preserves the predictive behavior of the original model while allowing end users to provide only a single, intuitive creditworthiness indicator.

```python
def credit_score_to_ext_sources(credit_score):
    
    credit_score = max(
    300,
    min(850, credit_score)
)

    score_normalized = (
    credit_score - 300
) / 550
    
    ext_source_1 = score_normalized * 0.95
    
    ext_source_2 = score_normalized * 0.98
    
    ext_source_3 = score_normalized
    
    ext_source_mean = (
        ext_source_1 +
        ext_source_2 +
        ext_source_3
    ) / 3
    
    return {
        "EXT_SOURCE_1": round(ext_source_1, 4),
        "EXT_SOURCE_2": round(ext_source_2, 4),
        "EXT_SOURCE_3": round(ext_source_3, 4),
        "EXT_SOURCE_MEAN": round(ext_source_mean, 4)
    }
```

## Enhanced Credit Score Calibration

The original machine learning models rely heavily on external bureau variables such as EXT_SOURCE_1, EXT_SOURCE_2, and EXT_SOURCE_3. These variables are among the most influential predictors of default risk but are not directly available to end users.

To support deployment without access to real credit bureau integrations, CreditWise AI generates synthetic EXT_SOURCE values using the customer's credit score.

### Credit Score Categories

The calibration follows widely used credit score classifications:

* Poor: 300–579
* Fair: 580–669
* Good: 670–739
* Very Good: 740–799
* Excellent: 800–850

Each category is assigned a realistic EXT_SOURCE range representing increasing levels of creditworthiness.

### Piecewise Calibration Strategy

Instead of applying a single linear transformation across the entire credit score range, the system uses a piecewise mapping approach.

Within each credit category:

* Higher scores produce higher EXT_SOURCE values.
* Lower scores produce lower EXT_SOURCE values.
* Linear interpolation preserves score granularity.

This allows two customers within the same category to receive different creditworthiness estimates while maintaining realistic risk bands.

### Simulating Multiple Bureau Signals

The original dataset contains three independent bureau-derived variables:

* EXT_SOURCE_1
* EXT_SOURCE_2
* EXT_SOURCE_3

To better approximate real-world behavior, all three values are generated from the same underlying credit score but include small deterministic offsets.

This preserves the strong correlation expected between bureau signals while avoiding unrealistic identical values.

### Benefits

This approach provides:

* Realistic credit score interpretation
* Consistent and reproducible predictions
* Improved approximation of external bureau information
* Better alignment with real-world lending practices
* Compatibility with the original machine learning models

The resulting EXT_SOURCE variables serve as a practical substitute for unavailable bureau data while preserving the predictive behavior learned during training.


## Credit Score to EXT_SOURCE Generation

The following function converts a user's Credit Score (CIBIL Score) into synthetic bureau features.

The implementation combines:

- Industry-standard credit score categories
- Piecewise calibration
- Linear interpolation within each category
- Small deterministic offsets between bureau variables

This produces realistic EXT_SOURCE values while maintaining deterministic and reproducible behavior.

```python
def credit_score_to_ext_sources(credit_score):

    credit_score = max(
        300,
        min(850, credit_score)
    )

    bands = [
    (300, 579, 0.05, 0.30, "Poor"),
    (580, 669, 0.30, 0.55, "Fair"),
    (670, 739, 0.55, 0.75, "Good"),
    (740, 799, 0.75, 0.90, "Very Good"),
    (800, 850, 0.90, 1.00, "Excellent")
]

    for score_min, score_max, ext_min, ext_max, category in bands:

        if score_min <= credit_score <= score_max:

            position = (
                credit_score - score_min
            ) / (
                score_max - score_min
            )

            base_score = (
                ext_min +
                position * (ext_max - ext_min)
            )

            ext_source_1 = max(
                0,
                min(1, base_score * 0.98)
            )

            ext_source_2 = max(
                0,
                min(1, base_score)
            )

            ext_source_3 = max(
                0,
                min(1, base_score * 1.02)
            )

            ext_source_mean = (
                ext_source_1 +
                ext_source_2 +
                ext_source_3
            ) / 3

            return {
                "EXT_SOURCE_1": round(ext_source_1, 4),
                "EXT_SOURCE_2": round(ext_source_2, 4),
                "EXT_SOURCE_3": round(ext_source_3, 4),
                "EXT_SOURCE_MEAN": round(ext_source_mean, 4),
                "CREDIT_PROFILE_CATEGORY": category
            }
```

## Recommendation Engine Design

Prediction alone does not provide actionable guidance to users.

To make CreditWise AI more useful and user-centric, a recommendation engine is introduced.

The recommendation engine analyzes customer characteristics and risk factors to generate personalized suggestions that may improve creditworthiness and reduce default risk.

Recommendations are generated using business rules derived from common lending practices and credit risk principles.

The objective is not only to predict risk but also to explain potential improvement opportunities.

### Recommendation Categories

Recommendations are generated from multiple perspectives:

1. Credit Score Improvement
2. Employment Stability
3. Debt Burden Reduction
4. Loan Affordability
5. Financial Capacity
6. General Credit Health

A customer may receive multiple recommendations depending on their profile.

```python
#v1
def generate_recommendations(
    age,
    income,
    employment_years,
    credit_score,
    credit_amount,
    annuity_amount
):

    recommendations = []

    # Credit score

    if credit_score < 580:
        recommendations.append(
            "Improve your credit score by reducing outstanding debt and maintaining timely repayments."
        )

    elif credit_score < 670:
        recommendations.append(
            "A moderate increase in credit score could significantly improve loan approval chances."
        )

    # Employment

    if employment_years < 2:
        recommendations.append(
            "Longer employment history may improve lender confidence and reduce perceived risk."
        )

    # Debt burden

    debt_ratio = annuity_amount / income

    if debt_ratio > 0.30:
        recommendations.append(
            "Current repayment obligations appear high relative to income. Consider reducing requested loan size."
        )

    # Large loan request

    if credit_amount > income * 5:
        recommendations.append(
            "Requested credit amount is substantially higher than annual income and may increase lending risk."
        )

    # Young applicant

    if age < 25:
        recommendations.append(
            "Building a longer credit history may strengthen future credit applications."
        )

    if len(recommendations) == 0:
        recommendations.append(
            "Customer profile appears financially healthy with no major risk indicators detected."
        )

    return recommendations
```

### Strategy-Aware Recommendations

Recommendation generation is influenced by the selected lending strategy.

Different institutions may have different risk tolerances:

- Risk Sensitive → Conservative recommendations
- Balanced → Moderate recommendations
- Performance Optimized → More permissive recommendations

This allows CreditWise AI to simulate multiple lending philosophies while keeping the underlying prediction model unchanged.

```python
#v3
def generate_recommendations(
    age,
    income,
    employment_years,
    credit_score,
    credit_amount,
    annuity_amount,
    strategy="balanced"
):

    recommendations = []

    # Strategy thresholds

    if strategy == "risk_sensitive":

        credit_score_threshold = 700
        debt_ratio_threshold = 0.25

    elif strategy == "performance_optimized":

        credit_score_threshold = 620
        debt_ratio_threshold = 0.40

    else:  # balanced

        credit_score_threshold = 670
        debt_ratio_threshold = 0.30

    # Credit Score

    if credit_score < credit_score_threshold - 90:

        recommendations.append({
            "category": "Credit Score",
            "priority": "High",
            "recommendation":
            "Improve your credit score by reducing outstanding debt and maintaining timely repayments."
        })

    elif credit_score < credit_score_threshold:

        recommendations.append({
            "category": "Credit Score",
            "priority": "Medium",
            "recommendation":
            "A moderate increase in credit score could significantly improve loan approval chances."
        })

    # Employment Stability

    if employment_years < 2:

        recommendations.append({
            "category": "Employment Stability",
            "priority": "Medium",
            "recommendation":
            "Longer employment history may improve lender confidence and reduce perceived risk."
        })

    # Debt Burden

    debt_ratio = annuity_amount / income

    if debt_ratio > debt_ratio_threshold:

        recommendations.append({
            "category": "Debt Burden",
            "priority": "High",
            "recommendation":
            "Current repayment obligations appear high relative to income. Consider reducing requested loan size."
        })

    # Loan Affordability

    if credit_amount > income * 5:

        recommendations.append({
            "category": "Loan Affordability",
            "priority": "High",
            "recommendation":
            "Requested credit amount is substantially higher than annual income and may increase lending risk."
        })

    # Credit History

    if age < 25:

        recommendations.append({
            "category": "Credit History",
            "priority": "Low",
            "recommendation":
            "Building a longer credit history may strengthen future credit applications."
        })

    # Healthy Profile

    if len(recommendations) == 0:

        recommendations.append({
            "category": "General",
            "priority": "Info",
            "recommendation":
            "Customer profile appears financially healthy with no major risk indicators detected."
        })

    # Sort recommendations by priority

    priority_order = {
        "High": 1,
        "Medium": 2,
        "Low": 3,
        "Info": 4
    }

    recommendations = sorted(
        recommendations,
        key=lambda x: priority_order[x["priority"]]
    )

    return {
        "strategy": strategy,
        "recommendation_count": len(recommendations),
        "recommendations": recommendations
    }
```

```python
def generate_recommendations(
    age,
    income,
    employment_years,
    credit_score,
    credit_amount,
    annuity_amount,
    education_level,
    family_status,
    housing_type,
    children_count,
    income_type,
    strategy="balanced"
):
    recommendations = []

    # Strategy thresholds

    if strategy == "risk_sensitive":

        credit_score_threshold = 700
        debt_ratio_threshold = 0.25

    elif strategy == "performance_optimized":

        credit_score_threshold = 620
        debt_ratio_threshold = 0.40

    else:  # balanced

        credit_score_threshold = 670
        debt_ratio_threshold = 0.30

    # Credit Score

    if credit_score < credit_score_threshold - 90:

        recommendations.append({
            "category": "Credit Score",
            "priority": "High",
            "recommendation":
            "Improve your credit score by reducing outstanding debt and maintaining timely repayments."
        })

    elif credit_score < credit_score_threshold:

        recommendations.append({
            "category": "Credit Score",
            "priority": "Medium",
            "recommendation":
            "A moderate increase in credit score could significantly improve loan approval chances."
        })

    # Employment Stability

    if employment_years < 2:

        recommendations.append({
            "category": "Employment Stability",
            "priority": "Medium",
            "recommendation":
            "Longer employment history may improve lender confidence and reduce perceived risk."
        })

    # Debt Burden

    debt_ratio = annuity_amount / income

    if debt_ratio > debt_ratio_threshold:

        recommendations.append({
            "category": "Debt Burden",
            "priority": "High",
            "recommendation":
            "Current repayment obligations appear high relative to income. Consider reducing requested loan size."
        })

    # Loan Affordability

    if credit_amount > income * 5:

        recommendations.append({
            "category": "Loan Affordability",
            "priority": "High",
            "recommendation":
            "Requested credit amount is substantially higher than annual income and may increase lending risk."
        })

    # Credit History

    if age < 25:

        recommendations.append({
            "category": "Credit History",
            "priority": "Low",
            "recommendation":
            "Building a longer credit history may strengthen future credit applications."
        })

    # Healthy Profile

    if len(recommendations) == 0:

        recommendations.append({
            "category": "General",
            "priority": "Info",
            "recommendation":
            "Customer profile appears financially healthy with no major risk indicators detected."
        })
    
    # Education Level

    if education_level in [
    "Lower secondary",
    "Secondary / secondary special"
]:

        recommendations.append({
            "category": "Education",
            "priority": "Low",
            "recommendation":
            "Higher educational qualifications may improve long-term income growth and financial stability."
        })

    # Family Status

    if family_status in [
    "Single / not married",
    "Separated"
]:

        recommendations.append({
            "category": "Financial Planning",
            "priority": "Low",
            "recommendation":
            "Maintaining stable savings and repayment habits can strengthen your financial profile."
        })

    # Housing Type

    if housing_type not in [
    "House / apartment"
]:

        recommendations.append({
            "category": "Housing Stability",
            "priority": "Low",
            "recommendation":
            "Stable residential arrangements may improve overall lending confidence."
        })

    # Children Count

    if children_count >= 3:

        recommendations.append({
            "category": "Household Expenses",
            "priority": "Medium",
            "recommendation":
            "Large household expenses may impact repayment capacity. Consider maintaining an emergency fund."
        })

    # Income Type

    if income_type in [
    "Unemployed",
    "Student",
    "Maternity leave"
]:

        recommendations.append({
            "category": "Income Stability",
            "priority": "High",
            "recommendation":
            "Stable and recurring income sources generally improve loan eligibility and reduce lending risk."
        })
```

## Improvement Simulator

The Improvement Simulator allows customers to evaluate how changes in their financial profile may affect predicted default risk.

- Credit Score
- Income
- Loan Amount

### Improvement Simulator Architecture

The Improvement Simulator evaluates how changes in customer characteristics may affect predicted default risk.

The simulator operates by modifying selected customer attributes and re-running the complete prediction pipeline.

This enables what-if analysis without retraining the machine learning model.

Two simulation modes are supported:

- Manual Scenario Simulation
- Recommended Improvement Simulation

The simulator reuses the same trained model, feature engineering logic, and risk categorization framework used during normal prediction.

```python
def create_simulated_profile(
    customer_profile,
    updates
):
    
    simulated_profile = customer_profile.copy()

    for feature, value in updates.items():

        simulated_profile[feature] = value

    return simulated_profile
```

```python
sample_profile = {
    "AGE_YEARS": 30,
    "INCOME_TOTAL": 500000,
    "EMPLOYMENT_YEARS": 4,
    "CREDIT_SCORE": 650
}

improved_profile = create_simulated_profile(
    sample_profile,
    {
        "CREDIT_SCORE": 750
    }
)

print(improved_profile)
```

### Credit Score Improvement Simulator

The first simulator evaluates the impact of improving a customer's credit score.

Rather than modifying the machine learning model, the simulator creates alternative customer scenarios and compares them against the current profile.

This enables users to understand how improving creditworthiness may influence future lending outcomes.

```python
def simulate_credit_score_improvement(
    customer_profile,
    improvement_points=50
):

    current_score = customer_profile["CREDIT_SCORE"]

    improved_score = min(
        850,
        current_score + improvement_points
    )

    improved_profile = create_simulated_profile(
        customer_profile,
        {
            "CREDIT_SCORE": improved_score
        }
    )

    return {
        "current_credit_score": current_score,
        "improved_credit_score": improved_score,
        "score_improvement": improved_score - current_score,
        "current_credit_profile":
            credit_score_to_ext_sources(
                current_score
            )["CREDIT_PROFILE_CATEGORY"],
        "improved_credit_profile":
            credit_score_to_ext_sources(
                improved_score
            )["CREDIT_PROFILE_CATEGORY"]
    }
```

### Multi-Scenario Simulation

A single recommendation often does not capture the full range of possible customer improvements.

The Multi-Scenario Simulator automatically generates several realistic improvement scenarios and compares them against the customer's current profile.

This allows users to evaluate multiple financial strategies before making decisions.

Example scenarios include:

- Credit Score Improvement
- Income Growth
- Loan Reduction
- Employment Stability Improvement

The simulator forms the foundation of the What-If Analysis Dashboard.

```python
def simulate_multiple_scenarios(
    customer_profile
):

    scenarios = {}

    # Current Profile

    scenarios["Current"] = customer_profile.copy()

    # Credit Score +50

    scenarios["Credit Score +50"] = create_simulated_profile(
        customer_profile,
        {
            "CREDIT_SCORE": min(
                850,
                customer_profile["CREDIT_SCORE"] + 50
            )
        }
    )

    # Credit Score +100

    scenarios["Credit Score +100"] = create_simulated_profile(
        customer_profile,
        {
            "CREDIT_SCORE": min(
                850,
                customer_profile["CREDIT_SCORE"] + 100
            )
        }
    )

    # Income +20%

    scenarios["Income +20%"] = create_simulated_profile(
        customer_profile,
        {
            "INCOME_TOTAL":
            customer_profile["INCOME_TOTAL"] * 1.20
        }
    )

    # Loan Amount -20%

    scenarios["Loan Amount -20%"] = create_simulated_profile(
        customer_profile,
        {
            "AMT_CREDIT":
            customer_profile["AMT_CREDIT"] * 0.80
        }
    )

    # Employment +2 Years

    scenarios["Employment +2 Years"] = create_simulated_profile(
        customer_profile,
        {
            "EMPLOYMENT_YEARS":
            customer_profile["EMPLOYMENT_YEARS"] + 2
        }
    )

    return scenarios
```

```python
sample_customer = {
    "CREDIT_SCORE": 620,
    "INCOME_TOTAL": 500000,
    "AMT_CREDIT": 1000000,
    "EMPLOYMENT_YEARS": 4
}

scenarios = simulate_multiple_scenarios(
    sample_customer
)

for name, scenario in scenarios.items():

    print("\n")
    print(name)
    print(scenario)
```

### Guided Improvement Plans

Many users may not know which financial changes would have the greatest impact on their risk profile.

Guided Improvement Plans provide predefined improvement strategies based on common financial goals.

These plans modify multiple customer attributes simultaneously and represent realistic improvement pathways.

```python
def generate_guided_plans(
    customer_profile
):

    plans = {}

    # Conservative Plan

    plans["Conservative Plan"] = create_simulated_profile(
        customer_profile,
        {
            "CREDIT_SCORE":
                min(
                    850,
                    customer_profile["CREDIT_SCORE"] + 50
                ),

            "INCOME_TOTAL":
                customer_profile["INCOME_TOTAL"] * 1.10,

            "EMPLOYMENT_YEARS":
                customer_profile["EMPLOYMENT_YEARS"] + 1
        }
    )

    # Moderate Plan

    plans["Moderate Plan"] = create_simulated_profile(
        customer_profile,
        {
            "CREDIT_SCORE":
                min(
                    850,
                    customer_profile["CREDIT_SCORE"] + 100
                ),

            "INCOME_TOTAL":
                customer_profile["INCOME_TOTAL"] * 1.20,

            "AMT_CREDIT":
                customer_profile["AMT_CREDIT"] * 0.90,

            "EMPLOYMENT_YEARS":
                customer_profile["EMPLOYMENT_YEARS"] + 2
        }
    )

    # Aggressive Plan

    plans["Aggressive Plan"] = create_simulated_profile(
        customer_profile,
        {
            "CREDIT_SCORE":
                min(
                    850,
                    customer_profile["CREDIT_SCORE"] + 150
                ),

            "INCOME_TOTAL":
                customer_profile["INCOME_TOTAL"] * 1.30,

            "AMT_CREDIT":
                customer_profile["AMT_CREDIT"] * 0.80,

            "EMPLOYMENT_YEARS":
                customer_profile["EMPLOYMENT_YEARS"] + 3
        }
    )

    # Best Case

    plans["Best Case Scenario"] = create_simulated_profile(
        customer_profile,
        {
            "CREDIT_SCORE": 800,

            "INCOME_TOTAL":
                customer_profile["INCOME_TOTAL"] * 1.50,

            "AMT_CREDIT":
                customer_profile["AMT_CREDIT"] * 0.70,

            "EMPLOYMENT_YEARS":
                customer_profile["EMPLOYMENT_YEARS"] + 5
        }
    )

    return plans
```

```python
guided_plans = generate_guided_plans(
    sample_customer
)

for name, plan in guided_plans.items():

    print("\n")
    print(name)
    print(plan)
```

### Custom Scenario Builder

The Custom Scenario Builder allows users to create their own financial improvement or deterioration scenarios.

Users may specify:

- Credit Score Changes
- Income Changes
- Loan Amount Changes
- Employment History Changes

Both positive and negative adjustments are supported to enable realistic what-if analysis.

```python
def simulate_custom_scenario(
    customer_profile,

    credit_score_change=0,

    income_change_pct=0,

    loan_change_pct=0,

    employment_change_years=0
):

    updated_profile = create_simulated_profile(
        customer_profile,
        {
            "CREDIT_SCORE":
                max(
                    300,
                    min(
                        850,
                        customer_profile["CREDIT_SCORE"]
                        + credit_score_change
                    )
                ),

            "INCOME_TOTAL":
                customer_profile["INCOME_TOTAL"]
                *
                (
                    1 + income_change_pct / 100
                ),

            "AMT_CREDIT":
                customer_profile["AMT_CREDIT"]
                *
                (
                    1 + loan_change_pct / 100
                ),

            "EMPLOYMENT_YEARS":
                max(
                    0,
                    customer_profile["EMPLOYMENT_YEARS"]
                    + employment_change_years
                )
        }
    )

    return updated_profile
```

```python
custom_scenario = simulate_custom_scenario(
    sample_customer,

    credit_score_change=75,

    income_change_pct=30,

    loan_change_pct=-15,

    employment_change_years=2
)

custom_scenario
```

### Master Simulator Engine

The Master Simulator Engine acts as a central orchestration layer for all simulation capabilities.

Rather than invoking individual simulation functions separately, the frontend or API can call a single function to generate:

- Quick Simulations
- Guided Improvement Plans
- Custom Scenarios

This design improves maintainability and simplifies frontend integration.

```python
def run_full_simulator(
    customer_profile,
    custom_scenario=None
):

    results = {}

    results["quick_scenarios"] = (
        simulate_multiple_scenarios(
            customer_profile
        )
    )

    results["guided_plans"] = (
        generate_guided_plans(
            customer_profile
        )
    )

    if custom_scenario is not None:

        results["custom_scenario"] = (
            simulate_custom_scenario(
                customer_profile,
                **custom_scenario
            )
        )

    return results
```

```python
simulation_results = run_full_simulator(
    sample_customer,
    custom_scenario={
        "credit_score_change": 75,
        "income_change_pct": 30,
        "loan_change_pct": -15,
        "employment_change_years": 2
    }
)

simulation_results.keys()
```
